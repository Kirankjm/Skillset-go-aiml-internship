```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV
df = pd.read_csv("hyperparameter_comparison.csv")

# Display summary statistics
print(df.describe())

# Plot accuracy vs. final loss
plt.figure(figsize=(8,5))
plt.scatter(df["Accuracy (%)"], df["Final Loss"], color="teal", s=80)
plt.title("Accuracy vs Final Loss Across Experiments")
plt.xlabel("Accuracy (%)")
plt.ylabel("Final Loss")
plt.grid(True)
plt.show()

# Optional: Bar chart for accuracy comparison
plt.figure(figsize=(8,5))
plt.bar(range(len(df)), df["Accuracy (%)"], color="skyblue")
plt.title("Model Accuracy Across Experiments")
plt.xlabel("Experiment Index")
plt.ylabel("Accuracy (%)")
plt.show()
```
