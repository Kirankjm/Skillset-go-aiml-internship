```python
#problem1 -> statics 
#ML Use: Feature scaling and normalization.
import numpy as np
from scipy import stats

data = np.array([41, 8, 16, 25, 3])
mean = np.mean(data)
median = np.median(data)
mode = stats.mode(data, keepdims=True).mode[0]
variance = np.var(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
#---> probabilistic problem
#Problem 2 — Bayes’ Theorem
#ML Use: Naive Bayes Classifier.
# P(spam) = 0.2, P(predicted_spam|spam) = 0.9, P(predicted_spam|not_spam) = 0.1
P_spam = 0.2
P_pred_spam_given_spam = 0.9
P_pred_spam_given_not_spam = 0.1
P_not_spam = 1 - P_spam

P_spam_given_pred_spam = (P_pred_spam_given_spam * P_spam) / (
    (P_pred_spam_given_spam * P_spam) + (P_pred_spam_given_not_spam * P_not_spam)
)

print("P(spam | predicted_spam):", round(P_spam_given_pred_spam, 2))

```

```python
# ---> Linear Algebra
# Problem 3 — Matrix Multiplication
#ML Use: Neural‑network forward propagation.
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])
result = np.dot(A, B)
print(result)
```

```python
#Problem 4 — Eigenvalues and Eigenvectors
#ML Use: Principal Component Analysis (PCA).
M = np.array([[2, 1], [1, 2]])
vals, vecs = np.linalg.eig(M)
print("Eigenvalues:", vals)
print("Eigenvectors:\n", vecs)
```

```python
# --> Calculus
# Problem 5 — Derivative of a Function
#ML Use: Gradient computation for optimization.
import sympy as sp

x = sp.Symbol('x')
f = x**2 + 3*x + 2
derivative = sp.diff(f, x)
print("Derivative:", derivative)
```

```python
#Problem 6 — Gradient Descent Step
#ML Use: Parameter updates in gradient descent.
def gradient_descent_step(x, lr):
    grad = 2 * (x - 3)  # derivative of (x - 3)^2
    x_new = x - lr * grad
    return x_new

x = 0
lr = 0.1
for i in range(5):
    x = gradient_descent_step(x, lr)
    print(f"Step {i+1}: x = {x:.4f}")
```
